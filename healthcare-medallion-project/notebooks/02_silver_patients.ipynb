```python
from pyspark.sql import functions as F

patients_bronze = spark.table(
    "healthcare.default.bronze_patients"
)

print("Bronze patient rows:", patients_bronze.count())
print("Bronze patient columns:", len(patients_bronze.columns))

display(patients_bronze)
```

```python
patients_bronze.printSchema()
```

```python
null_counts = patients_bronze.select([
    F.sum(
        F.when(F.col(c).isNull(), 1).otherwise(0)
    ).alias(c)
    for c in patients_bronze.columns
])

display(null_counts)
```

```python
duplicate_patients = (
    patients_bronze
    .groupBy("patient_id")
    .count()
    .filter(F.col("count") > 1)
)

print("Duplicate patient IDs:", duplicate_patients.count())

display(duplicate_patients)
```

```python
display(
    patients_bronze
    .groupBy("gender")
    .count()
    .orderBy("gender")
)
```

```python
display(
    patients_bronze.select(
        F.min("date_of_birth").alias("earliest_date_of_birth"),
        F.max("date_of_birth").alias("latest_date_of_birth"),
        F.min("registration_date").alias("earliest_registration_date"),
        F.max("registration_date").alias("latest_registration_date")
    )
)
```

```python
# Patient data quality validation

patients_checked = (
    patients_bronze

    # Standardize text fields
    .withColumn("first_name_clean", F.trim(F.col("first_name")))
    .withColumn("last_name_clean", F.trim(F.col("last_name")))
    .withColumn("email_clean", F.lower(F.trim(F.col("email"))))
    .withColumn("gender_clean", F.upper(F.trim(F.col("gender"))))
    .withColumn("insurance_provider_clean",
                F.trim(F.col("insurance_provider")))

    # Convert contact number to string for safe handling
    .withColumn(
        "contact_number_clean",
        F.col("contact_number").cast("string")
    )

    # Validation flags
    .withColumn(
        "valid_patient_id",
        F.col("patient_id").isNotNull()
    )

    .withColumn(
        "valid_gender",
        F.col("gender_clean").isin("M", "F")
    )

    .withColumn(
        "valid_dates",
        (
            F.col("date_of_birth").isNotNull()
            &
            F.col("registration_date").isNotNull()
            &
            (F.col("date_of_birth") <= F.col("registration_date"))
        )
    )

    .withColumn(
        "valid_email",
        F.col("email_clean").rlike(
            r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
        )
    )
)

display(
    patients_checked.select(
        "patient_id",
        "gender",
        "gender_clean",
        "date_of_birth",
        "registration_date",
        "email",
        "email_clean",
        "valid_patient_id",
        "valid_gender",
        "valid_dates",
        "valid_email"
    )
)
```

```python
validation_summary = patients_checked.select(
    F.count("*").alias("total_records"),
    F.sum(
        F.when(~F.col("valid_patient_id"), 1).otherwise(0)
    ).alias("invalid_patient_id"),
    F.sum(
        F.when(~F.col("valid_gender"), 1).otherwise(0)
    ).alias("invalid_gender"),
    F.sum(
        F.when(~F.col("valid_dates"), 1).otherwise(0)
    ).alias("invalid_dates"),
    F.sum(
        F.when(~F.col("valid_email"), 1).otherwise(0)
    ).alias("invalid_email")
)

display(validation_summary)
```

```python
# Recreate Silver patients using the standardized gender values

silver_patients = (
    patients_clean
    .select(
        "patient_id",

        F.col("first_name_clean").alias("first_name"),

        F.col("last_name_clean").alias("last_name"),

        "gender",

        "date_of_birth",

        F.col("contact_number_clean").alias("contact_number"),

        F.trim(F.col("address")).alias("address"),

        "registration_date",

        F.col("insurance_provider_clean").alias("insurance_provider"),

        F.trim(F.col("insurance_number")).alias("insurance_number"),

        F.col("email_clean").alias("email"),

        # Data-quality validation
        "valid_patient_id",
        "valid_gender",
        "valid_dates",
        "valid_email",

        # Lineage
        "_batch_id",
        "_source_id",
        "_source_name",
        "_source_file_name",
        "_ingestion_timestamp",
        "_ingestion_date",
        "_record_hash"
    )

    .filter(
        F.col("valid_patient_id")
        & F.col("valid_gender")
        & F.col("valid_dates")
        & F.col("valid_email")
    )
)

print("Silver patient records:", silver_patients.count())
```

```python
silver_patients.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("healthcare.default.silver_patients")

print("silver_patients updated successfully.")
```
